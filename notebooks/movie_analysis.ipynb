```python
decade_ratings = movies.groupby("Decade")["IMDB_Rating"].mean()

plt.figure(figsize=(10, 5))

decade_ratings.plot(kind="line", marker="o")

plt.title("Average IMDb Rating by Decade")
plt.xlabel("Decade")
plt.ylabel("Average IMDb Rating")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

genre_counts.head(10).sort_values().plot(kind="barh")

plt.title("Top 10 Movie Genre")
plt.xlabel("number of movies")
plt.ylabel("Genre")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

director_counts.head(10).sort_values().plot(kind="barh")

plt.title("Top 10 Directors by number of movies")
plt.xlabel("number of movies")
plt.ylabel("Director")

plt.show()
```

```python
top_voted = movies.nlargest(10, "No_of_Votes")

plt.figure(figsize=(12, 6))

plt.barh(
    top_voted["Series_Title"],
    top_voted["No_of_Votes"]
)

plt.title("Top 10 Movies by Number of Votes")
plt.xlabel("Number of Votes")
plt.ylabel("Movie")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
top_gross = movies.nlargest(10, "Gross")

plt.figure(figsize=(12, 6))

plt.barh(
    top_gross["Series_Title"],
    top_gross["Gross"] / 1_000_000
)

plt.title("Top 10 Highest Grossing movies")
plt.xlabel("Gross revenue (million USD)")
plt.ylabel("Movie")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.hist(movies["IMDB_Rating"], bins=15, edgecolor = "black")

plt.title("distribution of IMDB ratings")
plt.xlabel("IMDB Rating")
plt.ylabel("number of movies")

plt.show()
```
